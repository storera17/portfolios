```python
# Import Libraries
import pandas as pd 
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_squared_error, accuracy_score, classification_report, confusion_matrix
  

from math import sqrt

df = pd.read_csv("energy_data.csv")
# df.head() - preview the data
```

<div style="color:white;display:fill;
            background-color:#2a2a9fab;font-size:200%;">
    <p style="padding: 4px;color:white;"><b>Prepare the Data for Regression ML</b></p>
</div>

```python
df.columns
```

```python
features = ['Age', 'MaritalStatus', 'DayNightConsumption',
       'IncomeLevel', 'DwellingArea', 'HasChildren', 'SolarRoof',
       'ShiftableLoad', 'RiskAttitude', 'AttitudeSustainability', 'Tariff']      
```

```python
target = ['AnnualConsumption']
```

```python
# Extract Features and Target ('overall_rating') Values into Separate Dataframes
X = df[features]
y = df[target]
```

```python
# Let us look at a typical row from our features:
X.iloc[2]
```

```python
# create dummy variables for categorical columsn
X = pd.get_dummies(X,drop_first=True)
X.head()
```

```python
# Let us look at a typical row from our features:
y
```

```python
# split data into training (70%) and testing (30%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=50)
```

<div style="color:white;display:fill;
            background-color:#2a2a9fab;font-size:200%;">
    <p style="padding: 4px;color:white;"><b>Linear Regressor</b></p>
</div>

```python
regressor = LinearRegression()
regressor.fit(X_train, y_train)
y_prediction = regressor.predict(X_test)
# y_prediction
```

```python
# Mean of the expected target value in test set
y_test.describe()
```

```python
# Linear regression accuracy using root mean square error
RMSE = sqrt(mean_squared_error(y_true = y_test, y_pred = y_prediction))
print(RMSE)
```

<div style="color:white;display:fill;
            background-color:#2a2a9fab;font-size:200%;">
    <p style="padding: 4px;color:white;"><b>Decision Tree Regressor</b></p>
</div>

```python
regressor = DecisionTreeRegressor()
regressor.fit(X_train, y_train)
y_prediction = regressor.predict(X_test)
```

```python
# Mean of the expected target value in test set
y_test.describe()
```

```python
# decision tree regression accuracy using root mean square error
RMSE = sqrt(mean_squared_error(y_true = y_test, y_pred = y_prediction))
print(RMSE)
```

<div style="color:white;display:fill;
            background-color:#2a2a9fab;font-size:200%;">
    <p style="padding: 4px;color:white;"><b>Random Forest Regressor</b></p>
</div>

```python
regressor = RandomForestRegressor()
regressor.fit(X_train, y_train)
y_prediction = regressor.predict(X_test)
```

```python
# Mean of the expected target value in test set
y_test.describe()
```

```python
# random forest regression accuracy using root mean square error
RMSE = sqrt(mean_squared_error(y_true = y_test, y_pred = y_prediction))
print(RMSE)
```

```python
# select predictors
feature_columns = [
    "Age",
    "MaritalStatus",
    "AnnualConsumption",
    "DayNightConsumption",
    "IncomeLevel",
    "DwellingArea",
    "HasChildren",
    "SolarRoof",
    "ShiftableLoad",
    "RiskAttitude",
    "AttitudeSustainability"
]

X = df[feature_columns].copy()
X.head()
```

```python
# Convert categorical variables before creating dummy variables
categorical_columns = [
    "MaritalStatus",
    "IncomeLevel",
    "DwellingArea",
    "ShiftableLoad",
    "AttitudeSustainability",
    "HasChildren",
    "SolarRoof",
    "RiskAttitude"
]

for column in categorical_columns:
    X[column] = X[column].astype("category")

X = pd.get_dummies(X, columns=categorical_columns, drop_first=True)

X.head()
```

```python
# select target
y = df["Tariff"]

y.head()
```

```python
# split data into training (70%) and testing (30%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=50)
```

<div style="color:white;display:fill;
            background-color:#2a2a9fab;font-size:200%;">
    <p style="padding: 4px;color:white;"><b>Decision Tree Classifier</b></p>
</div>

```python
# initialize the algorithm
dtree=DecisionTreeClassifier(random_state=50)

# Generate a new model using training data only
dtree.fit(X_train,y_train)

# predict the test data
y_dtree_pred = dtree.predict(X_test) 
```

```python
# output the evaluation of decision tree
print(accuracy_score(y_test, y_dtree_pred))
print(confusion_matrix(y_test, y_dtree_pred))  
print(classification_report(y_test, y_dtree_pred)) 
print(dtree.feature_importances_)
```

<div style="color:white;display:fill;
            background-color:#2a2a9fab;font-size:200%;">
    <p style="padding: 4px;color:white;"><b>Random Forest Classifier</b></p>
</div>

```python
# create a random forest with 10 trees
rforest=RandomForestClassifier(random_state=50)

# learn the model using the training sets
rforest.fit(X_train,y_train)

# use the generated model to predict for the testing set
y_rforest_pred=rforest.predict(X_test)
```

```python
# output the evaluation of random forest
print(accuracy_score(y_test, y_rforest_pred))
print(confusion_matrix(y_test, y_rforest_pred))  
print(classification_report(y_test, y_rforest_pred)) 
print(rforest.feature_importances_)
```

<div style="color:white;display:fill;
            background-color:#2a2a9fab;font-size:200%;">
    <p style="padding: 4px;color:white;"><b>Cross Validated Decision Tree Classifier</b></p>
</div>

```python
# Create the parameter grid based on the results of random search 
param_grid = {
    'max_depth': [5, 10, 15],
    'min_samples_leaf': [1, 5, 10],
    'min_samples_split': [1, 5, 10],    
}

# Create a based model
dtree_cv = DecisionTreeClassifier()
# Instantiate the grid search model
grid_search = GridSearchCV(estimator = dtree_cv, param_grid = param_grid, cv = 5, n_jobs = -1, verbose = 2)

# Fit the grid search to the data
grid_search.fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
# initialize the algorithm
dtree_best=DecisionTreeClassifier(random_state=50, max_depth=5, min_samples_leaf=5, min_samples_split=10)

# Generate a new model using training data only
dtree_best.fit(X_train,y_train)

# predict the test data
y_dtree_best_pred = dtree.predict(X_test) 
```

```python
# output the evaluation of decision tree
print(accuracy_score(y_test, y_dtree_best_pred))
print(confusion_matrix(y_test, y_dtree_best_pred))  
print(classification_report(y_test, y_dtree_best_pred)) 
print(dtree_best.feature_importances_)
```

<div style="color:white;display:fill;
            background-color:#2a2a9fab;font-size:200%;">
    <p style="padding: 4px;color:white;"><b>Cross Validated Random Forest Classifier</b></p>
</div>

```python
# Create the parameter grid based on the results of random search 
param_grid = {
    'n_estimators': [5, 10, 15],
    'max_depth': [10, 15, 20]
}

# Create a based model
rforest_cv = RandomForestClassifier()

# Instantiate the grid search model
grid_search = GridSearchCV(estimator = rforest_cv, param_grid = param_grid, cv = 5, n_jobs = -1, verbose = 2)

# Fit the grid search to the data
grid_search.fit(X_train, y_train)
```

```python
# output the best values for tuning parameters
grid_search.best_params_
```

```python
# create a random forest with 10 trees
rforest_best=RandomForestClassifier(random_state=50, n_estimators=15, max_depth = 15)

# learn the model using the training sets
rforest_best.fit(X_train,y_train)

# use the generated model to predict for the testing set
y_rforest_best_pred=rforest_best.predict(X_test)
```

```python
# output the evaluation of random forest
print(accuracy_score(y_test, y_rforest_best_pred))
print(confusion_matrix(y_test, y_rforest_best_pred))  
print(classification_report(y_test, y_rforest_best_pred)) 
print(rforest_best.feature_importances_)
```
